```python
# install Modules using: pip install -r requirements.txt

import time
import random
import paho.mqtt.client as mqtt

BROKER = "192.168.12.1"
PORT = 1883
TOPIC = "building/room/temperature"

MIN_TEMP = 15
MAX_TEMP = 30
STEPS = 20  # 20 seconds, one value per second

def on_connect(client, userdata, flags, rc):
    print("Connected:", rc)

client = mqtt.Client()
client.on_connect = on_connect

client.connect(BROKER, PORT, 60)
client.loop_start()

try:
    while True:
        # Increase from 15 to 30 in 20 seconds
        for i in range(STEPS + 1):
            base_temp = MIN_TEMP + (MAX_TEMP - MIN_TEMP) * i / STEPS
            temp = base_temp + random.uniform(-0.1, 0.1)
            message = f"{temp:.2f}"

            res = client.publish(TOPIC, message)

            if res.rc == mqtt.MQTT_ERR_SUCCESS:
                print(f"Sent to {TOPIC}: {message}")
            else:
                print(f"Publish error: {res.rc}")

            time.sleep(1)

        # Decrease from 30 to 15 in 20 seconds
        for i in range(STEPS + 1):
            base_temp = MAX_TEMP - (MAX_TEMP - MIN_TEMP) * i / STEPS
            temp = base_temp + random.uniform(-0.1, 0.1)
            message = f"{temp:.2f}"

            res = client.publish(TOPIC, message)

            if res.rc == mqtt.MQTT_ERR_SUCCESS:
                print(f"Sent to {TOPIC}: {message}")
            else:
                print(f"Publish error: {res.rc}")

            time.sleep(1)

except KeyboardInterrupt:
    print("\nStopped by user.")

finally:
    client.loop_stop()
    client.disconnect()
    print("Disconnected.")
```
