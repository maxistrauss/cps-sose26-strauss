```python
# install Modules using: pip install -r requirements.txt

import paho.mqtt.client as mqtt

BROKER = "192.168.12.1"
PORT = 1883
TOPIC1 = "building/room1/temperature"
TOPIC2 = "building/room2/temperature"
TOPIC3 = "building/room3/temperature"
TOPIC4 = "building/room4/temperature"
MESSAGE = "Hallo MQTT"

client = mqtt.Client()

# Optional: Verbindungsausgabe
def on_connect(client, userdata, flags, rc):
    print("Connected:", rc)

client.on_connect = on_connect

# Verbindung herstellen
client.connect(BROKER, PORT, 60)

# Nachricht senden
res1 = client.publish(TOPIC1, MESSAGE)
res2 = client.publish(TOPIC2, MESSAGE)
res3 = client.publish(TOPIC3, MESSAGE)
res4 = client.publish(TOPIC4, MESSAGE)

# Prüfen ob erfolgreich
status = res1.rc

if status == mqtt.MQTT_ERR_SUCCESS:
    print(f"Nachricht gesendet an Topic '{TOPIC}'")
else:
    print(f"Fehler beim Senden: {status}")

# Verbindung sauber schließen
client.disconnect()
```
